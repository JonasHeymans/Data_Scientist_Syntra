# Starter Notebook — Plotly

## 1 — Setup

```
# If needed:
# %pip install -q plotly pandas

import pandas as pd
import plotly
import plotly.express as px
import plotly.graph_objects as go

print("Plotly version:", plotly.__version__)
```

## 2 — Load Dataset
We use the built-in Gapminder dataset from Plotly.

```

df = px.data.gapminder()

assert df.shape[0] > 0
assert {"country", "year", "continent", "lifeExp", "pop", "gdpPercap"}.issubset(df.columns)

df.head()
```

## 3 — Basic Scatter Plot

Create a scatter plot for year 2007:
- x = GDP per capita
- y = life expectancy
- color by continent
- size by population
- title must be explicit


```
df_2007 = df[df["year"] == 2007].copy()

fig = px.scatter(
    df_2007,
    x="gdpPercap",
    y="lifeExp",
    color="continent",
    size="pop",
    hover_name="country",
    size_max=60,
    title="Life Expectancy vs GDP per Capita (2007)"
)

fig.update_layout(margin=dict(l=40, r=40, t=60, b=40))
fig.show()
```

## 4 — Line Chart Over Time

Plot life expectancy over time for Belgium, China, and Brazil.


```

countries = ["Belgium", "China", "Brazil"]
df_sel = df[df["country"].isin(countries)].copy()

fig = px.line(
    df_sel,
    x="year",
    y="lifeExp",
    color="country",
    title="Life Expectancy Over Time"
)

fig.update_layout(margin=dict(l=40, r=40, t=60, b=40))
fig.show()
```

## 5 — Aggregation: Average Life Expectancy per Continent

Aggregate by continent and year.
Plot the average life expectancy over time.

```
agg = (
    df.groupby(["continent", "year"], as_index=False)
      .agg(avg_lifeExp=("lifeExp", "mean"))
)

fig = px.line(
    agg,
    x="year",
    y="avg_lifeExp",
    color="continent",
    title="Average Life Expectancy per Continent"
)

fig.show()
```

## 6 — Faceting

Create small multiples of GDP per capita vs life expectancy by continent for 2007.


```
fig = px.scatter(
    df_2007,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    hover_name="country",
    facet_col="continent",
    facet_col_wrap=3,
    title="GDP vs Life Expectancy by Continent (2007)"
)

fig.update_layout(margin=dict(l=40, r=40, t=60, b=40))
fig.show()
```
