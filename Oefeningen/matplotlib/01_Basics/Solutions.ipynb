# Matplotlib — Les 1 Oefeningen (Fundamenten)

**Doel van deze notebook:** na deze oefeningen kan je zelfstandig nette, leesbare plots maken met Matplotlib (OOP-stijl).

**Regels voor vandaag**
- Werk met de **OOP-interface**: `fig, ax = plt.subplots()`
- Elke plot heeft minstens: **titel**, **x-label**, **y-label**
- Gebruik een **legenda** wanneer er meerdere lijnen zijn
- Houd je plots **leesbaar** (limieten, grid optioneel, geen rommel)

> Tip: In Jupyter is `plt.show()` meestal niet nodig, maar in scripts vaak wel.


```
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
```

## A) Basisplots

### Oefening A1: Sinus

**Spec**
Plot een sinusfunctie van 0-20, met 200 punten


```
x = np.linspace(0,20, 200)

fig, ax = plt.subplots()

ax.plot(x, np.sin(x))
```

### Oefening A2: Sinus met limieten

Plot dezelfde sinusfunctie als uit A1, maar zoom in op enkel de positieve waarden van x 2.5 tot 17.5

```
x = np.linspace(0,20, 200)

fig, ax = plt.subplots()

ax.plot(x, np.sin(x))
ax.set_xlim(2.5, 17.5)
ax.set_ylim(0, 1)
```

## B) Leesbaarheid: legenda en stijl

### Oefening B1 — Twee lijnen + legenda

- Maak `y1 = sin(x)`
- Maak `y2 = cos(x)`
- Plot beide op dezelfde axes en voorzie een legenda

(x uit vorige oefening behouden)



```
x = np.linspace(0,20, 200)

fig, ax = plt.subplots()

ax.plot(x, np.sin(x), label='sin(x)')
ax.plot(x, np.cos(x), label='cos(x)')

ax.set_title('sin(x) vs cos(x)')
ax.set_xlabel('x (radians)')
ax.set_ylabel('waarde')
ax.legend();


```

### Oefening B2 — Stijlparameters

**Spec**
- Gebruik opnieuw `y1` en `y2`.
- Plot de cos lijn in stippellijntjes

Challenge: plot de sinuslijn met een dikte van 10
Extra challenge: Voeg een horizontale lijn toe op y=0

```
x = np.linspace(0,20, 200)

fig, ax = plt.subplots()

ax.plot(x, np.sin(x), label='sin(x)', linewidth = 10)
ax.plot(x, np.cos(x), label='cos(x)', linestyle='--')

ax.set_title('sin(x) vs cos(x)')
ax.set_xlabel('x (radians)')
ax.set_ylabel('waarde')
ax.legend();

ax.axhline(0)

```

## C) Subplots: 2×2 layout

### Oefening C1 — 2×2 subplots

- Recreëer volgende plots
- Kijk wat `fig.tight_layout()` doet


```
fig, axes = plt.subplots(2,2)

axes[0,0].plot(x, np.sin(x), label='sin', color='red')
axes[0,0].set(title='sin(x)')
axes[0,0].legend()

axes[1,0].plot(x, np.cos(x), label='cos', color='green')
axes[1,0].set(title='cos(x)')
axes[1,0].legend()

axes[0,1].plot(x-1, np.sin(x), label='sin',color='blue')
axes[0,1].set(title='sin(x)')
axes[0,1].legend()

axes[1,1].plot(x-1, np.cos(x), label='cos',color='orange')
axes[1,1].set(title='cos(x)')
axes[1,1].legend()

fig.tight_layout()
```

## D) Output

### Oefening D1 —  Oplaan van visualisaties

Maak een cosinusplot en sla deze op als 'cos.png'

```
x = np.linspace(0,20, 200)

fig, ax = plt.subplots()

ax.plot(x, np.cos(x))

fig.savefig('cos.png')
```

### Oefening D2 — Visualisatie vanuit pandas

We maken een klein dataframe.

**Spec**
- Plot `sales` vs `day` als een line chart met pandas.

Callenge: Pas daarna titel en labels aan via Matplotlib (`ax.set(...)`).



```
day = np.arange(1, 26)
week = range(1,50,2)
sales = 100 + 3*day + rng.normal(0, 15, size=day.size)

df = pd.DataFrame({"day": day, "sales": sales, 'week':week})

df.head()
```

```
df['sales'].plot()
```

```
x = df['day']
y = df['sales']

plt.plot(x, y)
plt.title('Sales over 30 dagen')
plt.xlabel('Dag')
plt.ylabel('Sales')

```
