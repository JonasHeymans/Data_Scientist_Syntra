## Setup
Eerst importeren we de nodige libraries.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

## Oefening 1: Seaborn Styles

Seaborn heeft verschillende ingebouwde styles die je plots er professioneel uit laten zien.

**Opdracht:**
1. Maak een eenvoudige lineplot met `x = [1, 2, 3, 4, 5]` en `y = [2, 4, 6, 8, 10]`
2. Probeer verschillende seaborn styles uit: `'darkgrid'`, `'whitegrid'`, `'dark'`, `'white'`, `'ticks'`
3. Gebruik `sns.set_style()` om de style te veranderen


```python
# Data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
```

```python
# Verschillende styles proberen
styles = ['darkgrid', 'whitegrid', 'dark', 'white', 'ticks']

for style in styles:
    sns.set_style(style)
    plt.figure(figsize=(8, 4))
    plt.plot(x, y, marker='o', linewidth=2)
    plt.title(f'Style: {style}')
    plt.xlabel('X-axis')
    plt.ylabel('Y-axis')
    plt.show()
```

## Oefening 2: Built-in Datasets

Seaborn heeft verschillende ingebouwde datasets die handig zijn voor oefenen.

**Opdracht:**
1. Laad de 'tips' dataset met `sns.load_dataset('tips')`
2. Doe basic EDA

```python
# Dataset laden
tips = sns.load_dataset('tips')

# Eerste 10 rijen
print("Eerste 10 rijen:")
print(tips.head(10))
print("\n" + "="*50 + "\n")

# Kolomnamen
print("Kolomnamen:")
print(tips.columns.tolist())
print("\n" + "="*50 + "\n")

# Beschrijving
print("Statistische beschrijving:")
print(tips.describe())
```

## Oefening 3: Histogram en KDE Plot

Distribution plots tonen hoe data verdeeld is.

**Opdracht:**
1. Maak een histogram van de 'total_bill' kolom uit de tips dataset
2. Voeg een KDE (Kernel Density Estimate) toe met de parameter `kde=True`

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=tips, x='total_bill', kde=True, bins=30)
plt.title('Verdeling van Total Bill met KDE', fontsize=14)
plt.xlabel('Total Bill ($)')
plt.ylabel('Frequentie')
```

## Oefening 4: Scatterplot

Scatterplots tonen de relatie tussen twee numerieke variabelen.

**Opdracht:**
1. Maak een scatterplot met 'total_bill' op de x-as en 'tip' op de y-as
2. Kleur de punten op basis van 'time' (lunch/dinner)


```python
plt.figure(figsize=(10, 6))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time', s=100, alpha=0.7)
plt.title('Relatie tussen Total Bill en Tip', fontsize=14)
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.legend(title='Tijd')
```

## Oefening 5: Box Plot

Box plots tonen de verdeling van data en outliers.

**Opdracht:**
1. Maak een box plot van 'total_bill' voor elke dag van de week

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=tips, x='day', y='total_bill')
plt.title('Total Bill Verdeling per Dag', fontsize=14)
plt.xlabel('Dag van de Week')
plt.ylabel('Total Bill ($)')
```

## Oefening 6: Violin Plot

Violin plots combineren box plots met KDE plots.

**Opdracht:**
1. Maak een violin plot van 'total_bill' voor 'lunch' vs 'dinner' (time kolom)
2. Splits de violin plot op basis van 'smoker' met de `hue` parameter
4. Toon de violins naast elkaar

```python
plt.figure(figsize=(10, 6))
sns.violinplot(data=tips, x='time', y='total_bill', hue='smoker', split=True)
plt.title('Total Bill Verdeling: Lunch vs Dinner (gesplitst op Smoker)', fontsize=14)
plt.xlabel('Tijd')
plt.ylabel('Total Bill ($)')
plt.legend(title='Smoker')
```

## Oefening 7: Count Plot

Count plots tellen het aantal observaties in elke categorie.

**Opdracht:**
1. Maak onderstaande plot van de 'day' kolom
2. Kleur de bars op basis van 'sex'


```python
plt.figure(figsize=(10, 6))
sns.countplot(data=tips, x='day', hue='sex')
plt.title('Aantal Bezoekers per Dag (gesplitst op Geslacht)', fontsize=14)
plt.xlabel('Dag van de Week')
plt.ylabel('Aantal')
plt.xticks(rotation=45)
plt.legend(title='Geslacht');
```

## Oefening 8: Bar Plot met Error Bars

Bar plots tonen gemiddelde waarden met error bars (standaarddeviatie).

**Opdracht:**
1. Maak een bar plot van de gemiddelde 'tip' per dag
2. Voeg een horizontale lijn toe op y=3

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=tips, x='day', y='tip', estimator=np.mean)
plt.axhline(y=3, color='red', linestyle='--', linewidth=2, label='Target: $3')
plt.title('Gemiddelde Tip per Dag', fontsize=14)
plt.xlabel('Dag van de Week')
plt.ylabel('Gemiddelde Tip ($)')
plt.legend();
```

## Oefening 9: Heatmap - Correlatie Matrix

Heatmaps zijn perfect voor het visualiseren van correlaties tussen variabelen.

**Opdracht:**
1. Maak een correlatie heatmap voor de verschillende features, voeg annotaties en een color map toe.

```python
# Numerieke kolommen selecteren
numeric_cols = tips.select_dtypes(include=[np.number])

# Correlatie matrix berekenen
corr_matrix = numeric_cols.corr()

# Heatmap maken
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, 
            linewidths=1, square=True, fmt='.2f')
plt.title('Correlatie Matrix - Tips Dataset', fontsize=14)
```

## Oefening 10: Pairplot

Een pairplot toont alle mogelijke scatterplots tussen variabelen.

**Opdracht:**
1. Maak een pairplot van de tips dataset en kleur de plots op basis van 'time'.


```python
# Pairplot maken (dit kan even duren)
pairplot = sns.pairplot(tips, hue='time', palette='husl', diag_kind='kde', height=2.5)
pairplot.figure.suptitle('Pairplot van Tips Dataset', y=1.02, fontsize=16)
plt.show()
```

## Oefening 11: Line Plot met meerdere lijnen

Line plots zijn handig voor tijdreeksen of geordende data.

**Opdracht:**
1. Laad de 'flights' dataset met `sns.load_dataset('flights')`
2. Maak de onderstaande plot van 'passengers' over 'year'
3. Geef elke maand een eigen lijn.
4. Maak de figuur groter voor betere leesbaarheid

```python
# Dataset laden
flights = sns.load_dataset('flights')

# Eerste rijen bekijken
print("Eerste rijen van flights dataset:")
print(flights.head())
print("\n")

# Line plot maken
plt.figure(figsize=(14, 8))
sns.lineplot(data=flights, x='year', y='passengers', hue='month', marker='o')
plt.title('Aantal Passagiers per Jaar (per Maand)', fontsize=16)
plt.xlabel('Jaar')
plt.ylabel('Aantal Passagiers')
plt.legend(title='Maand', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## Oefening 12: Custom Color Palettes

Seaborn heeft verschillende color palettes die je kunt gebruiken.

**Opdracht:**
1. Maak de onderstaande plot van 'day' uit de tips dataset
2. Probeer verschillende color palettes uit:
   - 'Set2'
   - 'husl'
   - 'viridis'

```python
# Verschillende palettes proberen
palettes = ['Set2', 'husl', 'viridis']

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for i, palette in enumerate(palettes):
    sns.countplot(data=tips, x='day', hue='day', palette=palette, ax=axes[i])
    axes[i].set_title(f'Palette: {palette}', fontsize=12)
    axes[i].set_xlabel('Dag')
    axes[i].set_ylabel('Aantal')

plt.tight_layout()
```

## Bonus Oefening: Combineer alles!

**Opdracht:**
1. Laad de 'iris' dataset
2. Maak een figure met 4 subplots (2x2)
3. Plot in elk subplot een ander type visualisatie:
   - Subplot 1: Scatter plot van sepal_length vs sepal_width (gekleurd op species)
   - Subplot 2: Box plot van petal_length per species
   - Subplot 3: Violin plot van petal_width per species
   - Subplot 4: Heatmap van de correlatie matrix
4. Voeg titels toe aan alle subplots


```python
# Iris dataset laden
iris = sns.load_dataset('iris')

# Figure met 4 subplots maken
fig, axes = plt.subplots(2, 2, figsize=(14, 12))

# Subplot 1: Scatter plot
sns.scatterplot(data=iris, x='sepal_length', y='sepal_width', 
                hue='species', s=100, alpha=0.7, ax=axes[0, 0])
axes[0, 0].set_title('Sepal Length vs Sepal Width', fontsize=12, fontweight='bold')
axes[0, 0].set_xlabel('Sepal Length (cm)')
axes[0, 0].set_ylabel('Sepal Width (cm)')
axes[0, 0].legend(title='Species')

# Subplot 2: Box plot
sns.boxplot(data=iris, x='species', y='petal_length', ax=axes[0, 1])
axes[0, 1].set_title('Petal Length per Species', fontsize=12, fontweight='bold')
axes[0, 1].set_xlabel('Species')
axes[0, 1].set_ylabel('Petal Length (cm)')

# Subplot 3: Violin plot
sns.violinplot(data=iris, x='species', y='petal_width', ax=axes[1, 0])
axes[1, 0].set_title('Petal Width per Species', fontsize=12, fontweight='bold')
axes[1, 0].set_xlabel('Species')
axes[1, 0].set_ylabel('Petal Width (cm)')

# Subplot 4: Heatmap van correlatie matrix
iris_numeric = iris.select_dtypes(include=[np.number])
corr_matrix = iris_numeric.corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0,
            linewidths=1, square=True, fmt='.2f', ax=axes[1, 1])
axes[1, 1].set_title('Correlatie Matrix', fontsize=12, fontweight='bold')

# Layout aanpassen
plt.tight_layout()
plt.show()

print("\nIris dataset info:")
print(iris.describe())
```

## Extra: Alternatieve oplossing voor Bonus met FacetGrid

Een andere manier om multiple plots te maken is met FacetGrid:

```python
# FacetGrid voorbeeld
g = sns.FacetGrid(iris, col='species', height=4, aspect=1)
g.map(sns.scatterplot, 'sepal_length', 'sepal_width', alpha=0.7)
g.set_axis_labels('Sepal Length (cm)', 'Sepal Width (cm)')
g.figure.suptitle('Sepal Dimensions per Species', y=1.02, fontsize=14)
```
